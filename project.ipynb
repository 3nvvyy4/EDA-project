```python
#melting the internet_sales dataframe:
copy = pd.melt(internet_sales_df, id_vars=[
    'OrderDate',
    'ShipDate',
    'CustomerKey',
    'PromotionKey',
    'SalesTerritoryKey',
    'SalesOrderNumber',
    'SalesOrderLineNumber',
    'RevisionNumber',
    'OrderQuantity',
    'UnitPrice',
    'ExtendedAmount',
    'UnitPriceDiscountPct',
    'DiscountAmount',
    'ProductStandardCost',
    'TotalProductCost'
], var_name="CurrencyKey", value_vars=[6,19,29,36,98,100], value_name="SalesAmount")
```

```python
copy['CurrencyKey'].value_counts(dropna=False)
```

```python
#mapping the currency key to the currency name
currency_map = {
    6: "AUD",
    19: "CAD",
    29: "DEM",
    36: "EUR",
    98: "GBP",
    100: "USD"
}
copy['CurrencyKey'] = copy['CurrencyKey'].map(currency_map)
print(copy['CurrencyKey'].value_counts(dropna=False))
```
